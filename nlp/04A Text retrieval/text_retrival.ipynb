# Imports

```python
import numpy as np
import pandas as pd
```

# Load data using pandas

```python
query = pd.read_parquet('query.parquet.brotli')
answer = pd.read_parquet('answer.parquet.brotli')
```

# Ranking

Apply some dark magic to select 5 candidates to answer each query.

```python
rng = np.random.default_rng()
prediction = rng.choice(len(answer), size=(len(query), 5))
```

# Save predictions

Save predictions file and upload to Cats

```python
np.savez('predictions.npz', prediction)
```
